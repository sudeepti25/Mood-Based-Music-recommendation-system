```python
# Remove any unnamed/index columns created by CSV exports (e.g. 'Unnamed: 0')
unnamed_cols = [c for c in df.columns if str(c).lower().startswith('unnamed')]
if unnamed_cols:
    df.drop(columns=unnamed_cols, inplace=True)
    print('Dropped columns:', unnamed_cols)
else:
    print('No unnamed columns found')

# Show current columns
print('Columns now:', df.columns.tolist())
```

```python




def mood_mapper(valence, energy, danceability, tempo, loudness):
    # Happy Energetic (Party) - very upbeat
    if valence >=0.6 and energy >=0.6 and danceability >=0.6 and tempo >=110 and loudness >=-6:
        return "Happy / Energetic (High Beat Party)"
    
    # Happy Energetic (Pleasant) - high valence but softer
    elif valence >=0.6 and energy >=0.5 and -12 <= loudness <= -6:
        return "Happy / Energetic (Pleasant)"
    
    # Uplifting Positive (medium-high tempo & mid-high energy)
    elif 0.4 <= valence <= 0.7 and 0.4 <= energy <= 0.7 and 90 <= tempo <= 120:
        return "Uplifting / Positive"
    
    # Calm / Peaceful (low energy + slow tempo + soft loudness)
    elif valence >= 0.5 and energy <= 0.4 and danceability <= 0.5 and tempo <= 90 and loudness <= -12:
        return "Calm / Peaceful"
    
    # Relaxation / Safety (neutral valence, low energy, soft loudness)
    elif 0.3 <= valence <= 0.6 and energy <= 0.4 and loudness <= -12:
        return "Relaxation / Safety"
    
    # Balanced / Neutral
    elif 0.4 <= valence <= 0.6 and 0.4 <= energy <= 0.6:
        return "Balanced / Neutral"
    
    # Fallback → classify by dominant feature
    else:
        if energy <= 0.3:
            return "Low Energy / Calm"
        elif valence <= 0.3:
            return "Sad / Negative"
        elif energy >= 0.7 and tempo >= 120:
            return "High Energy / Excited"
        else:
            return "Mixed / Uncategorized"


# Apply function
df["MoodCategory"] = df.apply(
    lambda row: mood_mapper(row["valence"], row["energy"], row["danceability"], row["tempo"], row["loudness"]),
    axis=1
)

# Save updated dataset
df.to_csv("dataset_with_moods_full.csv", index=False)

# Show preview
print(df[["track_name", "valence", "energy", "danceability", "tempo", "loudness", "MoodCategory"]].head())
```

```python
merge_map = {
    "Low Energy / Calm": "Calm & Relaxed",
    "Calm / Peaceful": "Calm & Relaxed",
    "Relaxation / Safety": "Calm & Relaxed"
}

# Apply mapping
df["MoodCategory"] = df["MoodCategory"].replace(merge_map)

# Check updated counts
```

## Valence vs Energy — Mood Regions
This plot shows audio tracks positioned by `valence` (x) and `energy` (y).
Colored translucent rectangles indicate approximate emotion regions derived from the `mood_mapper` rules in this notebook (2D projection).
Points are colored by the assigned `MoodCategory`. The figure is also saved as `valence_energy_mood_regions.png`.

```python
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.patches import Rectangle

# Ensure MoodCategory exists (in case cells were run out of order)
if 'MoodCategory' not in df.columns:
    df['MoodCategory'] = df.apply(lambda row: mood_mapper(row['valence'], row['energy'], row['danceability'], row['tempo'], row['loudness']), axis=1)

sns.set(style='whitegrid')
fig, ax = plt.subplots(figsize=(11,8))

# Define approximate 2D regions (valence on x-axis 0..1, energy on y-axis 0..1).
# These are projections of the rules used by mood_mapper (we ignore tempo/loudness/danceability here).
regions = [
    {'name':'Happy / Energetic','xmin':0.6,'xmax':1.0,'ymin':0.6,'ymax':1.0,'color':'#ffdddd'},
    {'name':'High Energy / Excited','xmin':0.0,'xmax':1.0,'ymin':0.7,'ymax':1.0,'color':'#ffe6cc'},
    {'name':'Sad / Negative','xmin':0.0,'xmax':0.3,'ymin':0.0,'ymax':1.0,'color':'#e6e6ff'},
    {'name':'Calm & Relaxed','xmin':0.3,'xmax':1.0,'ymin':0.0,'ymax':0.4,'color':'#e6f7ff'},
    {'name':'Balanced / Neutral','xmin':0.4,'xmax':0.6,'ymin':0.4,'ymax':0.6,'color':'#e6ffe6'}
]

# Draw translucent rectangles for regions (draw larger/background regions first)
for r in regions:
    rect = Rectangle((r['xmin'], r['ymin']), r['xmax']-r['xmin'], r['ymax']-r['ymin'],
                     facecolor=r['color'], edgecolor='none', alpha=0.35, zorder=0)
    ax.add_patch(rect)
    # annotate the region in its center
    cx = (r['xmin'] + r['xmax']) / 2
    cy = (r['ymin'] + r['ymax']) / 2
    ax.text(cx, cy, r['name'], ha='center', va='center', fontsize=9, alpha=0.9, zorder=5)

# Scatter plot of tracks colored by MoodCategory
# Use a palette with enough distinct colors
unique_moods = df['MoodCategory'].astype(str).unique()
palette = sns.color_palette('tab10', n_colors=len(unique_moods))
mood_to_color = {m: palette[i % len(palette)] for i, m in enumerate(sorted(unique_moods))}
colors = df['MoodCategory'].map(mood_to_color)

ax.scatter(df['valence'], df['energy'], c=list(colors), s=30, edgecolor='k', linewidth=0.2, alpha=0.85, zorder=6)

# Create a manual legend for moods (keeps it tidy)
from matplotlib.lines import Line2D
legend_elements = [Line2D([0], [0], marker='o', color='w', label=m, markerfacecolor=mood_to_color[m], markersize=8, markeredgecolor='k', linewidth=0) for m in sorted(unique_moods)]
ax.legend(handles=legend_elements, title='MoodCategory', bbox_to_anchor=(1.02, 1), loc='upper left')

ax.set_xlim(-0.02, 1.02)
ax.set_ylim(-0.02, 1.02)
ax.set_xlabel('Valence (negative → positive)')
ax.set_ylabel('Energy (low → high)')
ax.set_title('Valence vs Energy — Mood regions (approx.)')
plt.tight_layout()
# Save figure
out_path = 'valence_energy_mood_regions.png'
fig.savefig(out_path, dpi=200)
print(f'Saved figure to {out_path}')
plt.show()
```

```python
# Train a Random Forest classifier with anti-overfitting techniques
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, f1_score
import joblib

# Features to use
features = ['valence','energy','danceability','tempo','loudness']
df_rf = df.copy()
df_rf = df_rf.dropna(subset=features + ['MoodCategory'])
X = df_rf[features].astype(float).values
y = df_rf['MoodCategory'].astype(str).values

# Encode labels
le = LabelEncoder()
y_enc = le.fit_transform(y)
print('Classes:', list(le.classes_))
print(f'Total samples: {len(X)}')

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(X, y_enc, test_size=0.2, random_state=42, stratify=y_enc)

# Scale features to reduce variance (helps with overfitting)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print('\n=== Training Random Forest with Regularization ===')

# Random Forest with anti-overfitting parameters:
# - max_depth: limit tree depth to prevent overfitting
# - min_samples_split: require more samples to split nodes
# - min_samples_leaf: require more samples in leaf nodes
# - max_features: use subset of features per tree (adds randomness)
# - n_estimators: reduced from 300 to avoid overfitting
rf = RandomForestClassifier(
    n_estimators=100,           # Reduced from 300
    max_depth=10,                # Limit tree depth (prevent overfitting)
    min_samples_split=20,        # Require at least 20 samples to split
    min_samples_leaf=10,         # Require at least 10 samples in each leaf
    max_features='sqrt',         # Use sqrt(n_features) per tree
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train_scaled, y_train)

# Evaluate on training set (to check overfitting)
y_train_pred = rf.predict(X_train_scaled)
train_acc = accuracy_score(y_train, y_train_pred)
train_f1 = f1_score(y_train, y_train_pred, average='weighted')

# Evaluate on test set
y_test_pred = rf.predict(X_test_scaled)
test_acc = accuracy_score(y_test, y_test_pred)
test_f1 = f1_score(y_test, y_test_pred, average='weighted')

print(f'\n📊 Results:')
print(f'Training Accuracy: {train_acc:.4f}')
print(f'Test Accuracy:     {test_acc:.4f}')
print(f'Accuracy Gap:      {train_acc - test_acc:.4f}')
print(f'\nTraining F1:       {train_f1:.4f}')
print(f'Test F1:           {test_f1:.4f}')
print(f'F1 Gap:            {train_f1 - test_f1:.4f}')

if (train_acc - test_acc) > 0.1:
    print('\n⚠️  Large gap detected - model may still be overfitting')
else:
    print('\n✅ Gap is reasonable - overfitting reduced!')

print('\n📈 Classification Report (Test Set):')
print(classification_report(y_test, y_test_pred, target_names=le.classes_))

print('\n📊 Confusion Matrix (Test Set):')
print(confusion_matrix(y_test, y_test_pred))

# Cross-validation to check generalization
print('\n🔄 Performing 5-Fold Cross-Validation...')
cv_scores = cross_val_score(rf, X_train_scaled, y_train, cv=5, scoring='accuracy')
print(f'CV Accuracy: {cv_scores.mean():.4f} (+/- {cv_scores.std():.4f})')

# Feature importance
print('\n🔍 Feature Importance:')
for feat, imp in zip(features, rf.feature_importances_):
    print(f'  {feat:15s}: {imp:.4f}')

# Save the trained model with scaler
out_file = 'random_forest_mood_model.joblib'
joblib.dump({
    'model': rf, 
    'label_encoder': le, 
    'scaler': scaler,
    'features': features
}, out_file)
print(f'\n✅ Saved improved model to {out_file}')
```

## Hyperparameter Tuning (Optional)
Run this cell if you want to find the optimal Random Forest parameters to minimize overfitting.
This uses GridSearchCV to test different combinations and find the best settings.

```python
# Optional: GridSearch to find best hyperparameters
from sklearn.model_selection import GridSearchCV

print('🔍 Searching for best hyperparameters...')
print('⏳ This may take a few minutes...\n')

# Define parameter grid to search
param_grid = {
    'n_estimators': [50, 100, 150],
    'max_depth': [5, 10, 15, None],
    'min_samples_split': [10, 20, 30],
    'min_samples_leaf': [5, 10, 15],
    'max_features': ['sqrt', 'log2']
}

# Create base model
rf_base = RandomForestClassifier(
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)

# GridSearch with cross-validation
grid_search = GridSearchCV(
    rf_base,
    param_grid,
    cv=3,  # 3-fold CV to save time
    scoring='f1_weighted',
    n_jobs=-1,
    verbose=1
)

# Fit
grid_search.fit(X_train_scaled, y_train)

# Best parameters
print('\n✅ Best parameters found:')
for param, value in grid_search.best_params_.items():
    print(f'  {param}: {value}')

print(f'\n📊 Best CV Score: {grid_search.best_score_:.4f}')

# Evaluate best model on test set
best_model = grid_search.best_estimator_
y_pred_best = best_model.predict(X_test_scaled)
test_acc_best = accuracy_score(y_test, y_pred_best)
test_f1_best = f1_score(y_test, y_pred_best, average='weighted')

print(f'Test Accuracy: {test_acc_best:.4f}')
print(f'Test F1:       {test_f1_best:.4f}')

# Save tuned model if better
if test_f1_best > test_f1:
    print('\n🎉 Tuned model is better! Saving...')
    joblib.dump({
        'model': best_model,
        'label_encoder': le,
        'scaler': scaler,
        'features': features
    }, 'random_forest_mood_model_tuned.joblib')
    print('Saved as random_forest_mood_model_tuned.joblib')
else:
    print('\n📝 Original model performs better or similar. Keeping it.')
```

## Random Forest Visualizations
Comprehensive visualizations of model performance, feature importance, and prediction analysis.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (16, 12)

# Create a figure with multiple subplots
fig = plt.figure(figsize=(18, 12))

# 1. Confusion Matrix Heatmap
ax1 = plt.subplot(2, 3, 1)
cm = confusion_matrix(y_test, y_test_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=le.classes_, yticklabels=le.classes_,
            cbar_kws={'label': 'Count'}, ax=ax1)
ax1.set_title('Confusion Matrix', fontsize=14, fontweight='bold')
ax1.set_xlabel('Predicted Mood', fontsize=11)
ax1.set_ylabel('Actual Mood', fontsize=11)
plt.setp(ax1.get_xticklabels(), rotation=45, ha='right')
plt.setp(ax1.get_yticklabels(), rotation=0)

# 2. Feature Importance Bar Chart
ax2 = plt.subplot(2, 3, 2)
feature_imp = pd.DataFrame({
    'Feature': features,
    'Importance': rf.feature_importances_
}).sort_values('Importance', ascending=True)
colors = plt.cm.viridis(np.linspace(0.3, 0.9, len(feature_imp)))
ax2.barh(feature_imp['Feature'], feature_imp['Importance'], color=colors)
ax2.set_title('Feature Importance', fontsize=14, fontweight='bold')
ax2.set_xlabel('Importance Score', fontsize=11)
ax2.grid(axis='x', alpha=0.3)
for i, v in enumerate(feature_imp['Importance']):
    ax2.text(v + 0.01, i, f'{v:.3f}', va='center', fontsize=10)

# 3. Training vs Test Accuracy Comparison
ax3 = plt.subplot(2, 3, 3)
metrics = ['Accuracy', 'F1 Score']
train_scores = [train_acc, train_f1]
test_scores = [test_acc, test_f1]
x = np.arange(len(metrics))
width = 0.35
bars1 = ax3.bar(x - width/2, train_scores, width, label='Training', color='#4CAF50', alpha=0.8)
bars2 = ax3.bar(x + width/2, test_scores, width, label='Test', color='#2196F3', alpha=0.8)
ax3.set_ylabel('Score', fontsize=11)
ax3.set_title('Training vs Test Performance', fontsize=14, fontweight='bold')
ax3.set_xticks(x)
ax3.set_xticklabels(metrics)
ax3.legend()
ax3.set_ylim([0.95, 1.0])
ax3.grid(axis='y', alpha=0.3)
# Add value labels on bars
for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        ax3.text(bar.get_x() + bar.get_width()/2., height,
                f'{height:.4f}', ha='center', va='bottom', fontsize=9)

# 4. Per-Class Accuracy
ax4 = plt.subplot(2, 3, 4)
from sklearn.metrics import precision_recall_fscore_support
precision, recall, f1, support = precision_recall_fscore_support(y_test, y_test_pred)
class_names = le.classes_
x_pos = np.arange(len(class_names))
bars = ax4.bar(x_pos, recall, color=plt.cm.Set3(np.linspace(0, 1, len(class_names))), alpha=0.8)
ax4.set_xlabel('Mood Category', fontsize=11)
ax4.set_ylabel('Recall Score', fontsize=11)
ax4.set_title('Per-Class Recall (Accuracy)', fontsize=14, fontweight='bold')
ax4.set_xticks(x_pos)
ax4.set_xticklabels(class_names, rotation=45, ha='right')
ax4.set_ylim([0.95, 1.0])
ax4.grid(axis='y', alpha=0.3)
for i, (bar, val) in enumerate(zip(bars, recall)):
    ax4.text(i, val + 0.002, f'{val:.3f}', ha='center', va='bottom', fontsize=9)

# 5. Class Distribution
ax5 = plt.subplot(2, 3, 5)
class_counts = pd.Series(y_test).value_counts().sort_index()
class_labels = [le.classes_[i] for i in class_counts.index]
colors_pie = plt.cm.Set3(np.linspace(0, 1, len(class_counts)))
wedges, texts, autotexts = ax5.pie(class_counts.values, labels=class_labels, autopct='%1.1f%%',
                                     colors=colors_pie, startangle=90)
ax5.set_title('Test Set Class Distribution', fontsize=14, fontweight='bold')
plt.setp(autotexts, size=9, weight='bold')
plt.setp(texts, size=9)

# 6. Cross-Validation Scores
ax6 = plt.subplot(2, 3, 6)
from sklearn.model_selection import cross_val_score
cv_scores_full = cross_val_score(rf, X_train_scaled, y_train, cv=5, scoring='accuracy')
fold_numbers = np.arange(1, len(cv_scores_full) + 1)
bars = ax6.bar(fold_numbers, cv_scores_full, color='#FF9800', alpha=0.8, edgecolor='black')
ax6.axhline(y=cv_scores_full.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {cv_scores_full.mean():.4f}')
ax6.set_xlabel('Fold Number', fontsize=11)
ax6.set_ylabel('Accuracy', fontsize=11)
ax6.set_title('5-Fold Cross-Validation Scores', fontsize=14, fontweight='bold')
ax6.set_ylim([0.995, 1.0])
ax6.legend()
ax6.grid(axis='y', alpha=0.3)
for i, (bar, val) in enumerate(zip(bars, cv_scores_full)):
    ax6.text(i + 1, val - 0.0005, f'{val:.4f}', ha='center', va='top', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.savefig('random_forest_performance_analysis.png', dpi=300, bbox_inches='tight')
print('✅ Saved comprehensive visualization to random_forest_performance_analysis.png')
plt.show()
```

```python
# Prediction Probability Distribution
fig, axes = plt.subplots(2, 4, figsize=(20, 10))
axes = axes.flatten()

# Get prediction probabilities
y_proba = rf.predict_proba(X_test_scaled)

for idx, class_name in enumerate(le.classes_):
    ax = axes[idx]
    
    # Get probabilities for this class
    class_probabilities = y_proba[:, idx]
    
    # Separate correct vs incorrect predictions
    correct_mask = y_test == idx
    correct_probs = class_probabilities[correct_mask]
    incorrect_probs = class_probabilities[~correct_mask]
    
    # Plot histograms
    ax.hist(correct_probs, bins=50, alpha=0.7, color='green', label='Correct Predictions', edgecolor='black')
    ax.hist(incorrect_probs, bins=50, alpha=0.7, color='red', label='Incorrect Predictions', edgecolor='black')
    
    ax.set_title(f'{class_name}', fontsize=11, fontweight='bold')
    ax.set_xlabel('Prediction Confidence', fontsize=9)
    ax.set_ylabel('Frequency', fontsize=9)
    ax.legend(fontsize=8)
    ax.grid(alpha=0.3)
    
    # Add mean confidence text
    if len(correct_probs) > 0:
        mean_conf = correct_probs.mean()
        ax.axvline(mean_conf, color='green', linestyle='--', linewidth=2, alpha=0.7)
        ax.text(0.98, 0.95, f'Mean: {mean_conf:.3f}', transform=ax.transAxes,
                ha='right', va='top', fontsize=9, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.suptitle('Prediction Confidence Distribution by Mood Category', fontsize=16, fontweight='bold', y=1.00)
plt.tight_layout()
plt.savefig('prediction_confidence_distribution.png', dpi=300, bbox_inches='tight')
print('✅ Saved prediction confidence distribution to prediction_confidence_distribution.png')
plt.show()
```
